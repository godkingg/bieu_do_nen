```python
import pandas as pd
import os
import glob

# === 1. Đọc dữ liệu master ===
master_path = "master_data.csv"

if os.path.exists(master_path):
    df_master = pd.read_csv(master_path)
    display(df_master.head())
else:
    print("⚠️ Không tìm thấy file master_data.csv")
```

```python
data_dir = "data"
file_paths = glob.glob(os.path.join(data_dir, "*.csv"))
print(f"Có {len(file_paths)} file CSV trong thư mục '{data_dir}':")
print(file_paths[:5])  # in thử 5 file đầu

# Đọc 1 file mẫu
if file_paths:
    sample = pd.read_csv(file_paths[0])
    print(f"\nXem file mẫu: {os.path.basename(file_paths[0])}")
    display(sample.tail(5))
```

```python
# Nếu bạn lưu code vào file my_functions.py
from top10gainer import calculate_price_change_from_local_files

# Hoặc copy hàm vào cell phía trên

df_changes = calculate_price_change_from_local_files(df_master)
display(df_changes.head(10))
```
